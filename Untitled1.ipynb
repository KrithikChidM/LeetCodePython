#https://leetcode.com/problems/count-servers-that-communicate/?envType=daily-question&envId=2026-11-08

```python

class Solution:
    def canCommunicate(self, grid, i, j, m, n):
        cnt = 0
        if grid[i][j]==1:
            for k in range(n):
                cnt += int(grid[i][k]==1)
                if cnt > 1:
                    return True
            for k in range(m):
                cnt += int(grid[k][j]==1)
                if cnt > 2:
                    return True
        return False

    def countServers(self, grid: list[list[int]]) -> int:
        cnt = 0
        m = len(grid)
        n = len(grid[0])
        for i in range(m):
            for j in range(n):
                if self.canCommunicate(grid,i,j,m,n):
                    cnt += 1
        return cnt
```

```python
class Solution:
    def countServers(self, grid: list[list[int]]) -> int:
        m = len(grid)
        n = len(grid[0])
        row = [0] * m
        col = [0] * n
        for i in range(m):
            for j in range(n):
                if grid[i][j] == 1:
                    row[i] += 1
                    col[j] += 1
        ans = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == 1 and (row[i] > 1 or col[j] > 1):
                    ans += 1
        return ans
```
